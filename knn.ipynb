```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv(r"C:\Users\hp\Desktop\ml_project\data\train_binary.csv")
x = df.iloc[:,0:25]
m = int(0.8*(x.shape[0]))
x = (x - x.mean(axis=0))/x.std(axis=0)
x = x[:m]
x = x.to_numpy()
y = df.iloc[:,25]
y = y[:m]
k = 200

x_test = df.iloc[-1,0:25]
x_test = (x_test - x_test.mean(axis=0))/x_test.std(axis=0)
x_test = x_test.to_numpy()
def distance(x,x_test):
    distance = np.linalg.norm(x - x_test,axis=1)
    return distance
def sort(x,y,distance,x_test,k):
    distance = distance(x,x_test)
    indices = np.argsort(distance)[:k]
    labels = np.array(y[indices])
    return labels
def predict(y,distance,k):
    distance = distance(x,x_test)
    indices = np.argsort(distance)[:k]
    pred = np.argmax(np.bincount(y[indices]))
    return pred
label = sort(x,y,distance,x_test,k)
pred =  predict(y,distance,k)
print(pred,label)
```
